# Homework 03 (Data Streaming): Theory

Nome Cognome: Luca Simonelli

### Algoritmo 1
>INPUT: a stream $S$, two parameters $\delta>0, \epsilon > 0$;  
>IN MEMORY: an integer matrix $H$ of counters $H_{k,j}, k \in\{1, \ldots , K\} , j \in \{ 1, \ldots , J\}$;  
>QUERYING OUTPUT: for any $1 \leq i \leq n$, estimation of ${F}[i]$ with $\hat{F}[i]$ given below;  
>  
>// Initialization of hash functions and memory space  
>$K \longleftarrow \lceil \log \tfrac{1}{\delta} \rceil$;   
>$J \longleftarrow \lceil \tfrac{e}{\epsilon} \rceil$;  
>// definition of the K hash functions  
>FOR $k = 1, \ldots , K$:  
>  $h_k :\mathbb{Z}_+ \to \{1, \ldots , J\}$;  
>ENDFOR  
>// initialization of memory space: set to $0$ the matrix of counters  
>$H_{k,j} =0$, $k = 1, \ldots , K$, $j = 1, \ldots , J$  
>  
>// Streaming Algorithm  
>WHILE $\ldots,a_l,\ldots$ in stream $S$:  
>  FOR $k \in \{1,\ldots, K\}$:  
>    $H_{k,h_k(a_l)} \longleftarrow H_{k,h_k(a_l)} + 1$;  
>  ENDFOR  
>ENDWHILE  
>  
>// Querying Algorithm  
>INPUT: an integer $i$;  
>RETURN $\hat{F}[i] = \min_{k=1,\ldots,K} H_{k,h_k(i)}$.  
>
---

# Relazioni iniziali, 10pts

>
Con le notazioni dell'Esercizio 1, dimostrare che
>
- **ovvio:** per ogni $i \in \{1, \ldots, n\}$, $\hat{F}[i] \geq F[i]$.

---
<!-- Inizia qui la risposta -->
Per ogni oggetto $i$ nello stream e per ogni funzione hash $k \in \{1, \dots, K\}$, il contatore corrispondente nella matrice $H$ è definito come:
$$ H_{k, h_k(i)} = F[i] + \sum_{j \neq i, h_k(j)=h_k(i)} F[j] $$
Poiché le frequenze $F[j]$ sono per definizione non negative, la somma $\sum F[j]$ è sempre $\geq 0$. Di conseguenza:
$$ H_{k, h_k(i)} \geq F[i] \quad \forall k $$
Dato che la stima finale è il minimo tra tutti i $K$ contatori:
$$ \hat{F}[i] = \min_{k} H_{k, h_k(i)} \implies \hat{F}[i] \geq F[i] $$
<!-- Termina qui la risposta -->
---

>
- **meno facile:** per ogni $i \in \{1, \ldots, n\}$ e $k \in \{1, \ldots, K\}$:
$$
{E} [ H_{k,h_k(i)} ] \leq  F[i] + \frac{\epsilon}{e} (t - F[i]).
$$

---
<!-- Inizia qui la risposta -->
Sia $X_{k,j}$ una variabile aleatoria indicatrice che vale 1 se $h_k(j) = h_k(i)$ e 0 altrimenti. Per l'ipotesi di uniformità delle funzioni hash, $P(X_{k,j}=1) = \frac{1}{J}$.
Il valore del contatore può essere scritto come:
$$ H_{k, h_k(i)} = F[i] + \sum_{j \neq i} F[j] X_{k,j} $$
Passando al valore atteso e usando la linearità:
$$ E[H_{k, h_k(i)}] = F[i] + \sum_{j \neq i} F[j] E[X_{k,j}] = F[i] + \sum_{j \neq i} F[j] \frac{1}{J} $$
Sapendo che $\sum_{j \neq i} F[j] = t - F[i]$ e che $J = \lceil \frac{e}{\epsilon} \rceil$, allora $\frac{1}{J} \leq \frac{\epsilon}{e}$. Sostituendo otteniamo:
$$ E[H_{k, h_k(i)}] \leq F[i] + \frac{\epsilon}{e} (t - F[i]) $$
<!-- Termina qui la risposta -->

# Controllo probabilistico, 15pts
>
Utilizzando le relazione viste sopra, l'indipendenza delle funzione hash e la disuguaglianza di Markov dimostrare che
$$
P ( F[i] \leq \hat{F}[i] \leq F[i]+ \epsilon t) \geq 1-\delta.
$$

---
<!-- Inizia qui la risposta -->
Definiamo l'errore sulla riga $k$ come $E_k = H_{k, h_k(i)} - F[i]$. Dal punto precedente sappiamo che $E[E_k] \leq \frac{\epsilon}{e} t$.
Applichiamo la \textbf{disuguaglianza di Markov} per la probabilità che l'errore superi la soglia $\epsilon t$:
$$ P(E_k \geq \epsilon t) = P(E_k \geq e \cdot \frac{\epsilon}{e} t) \leq \frac{E[E_k]}{\epsilon t} \leq \frac{\frac{\epsilon}{e} t}{\epsilon t} = \frac{1}{e} $$
La stima finale $\hat{F}[i]$ fallisce solo se tutti i $K$ contatori superano la soglia. Data l'indipendenza delle funzioni hash:
$$ P(\hat{F}[i] - F[i] \geq \epsilon t) = P(\forall k, E_k \geq \epsilon t) \leq \left( \frac{1}{e} \right)^K $$
Sostituendo $K = \lceil \ln(1/\delta) \rceil$, otteniamo:
$$ P(\hat{F}[i] - F[i] \geq \epsilon t) \leq \left( \frac{1}{e} \right)^{\ln(1/\delta)} = \delta $$
Pertanto, la probabilità complementare (che la stima sia corretta entro l'errore $\epsilon t$) è:
$$ P(\hat{F}[i] \leq F[i] + \epsilon t) \geq 1 - \delta $$
